```r
display_markdown(" # Kaplan Meier: Todos os resultados")
display_markdown(" ## Valores das exposições de uma assinatura específica")
display_markdown(" ##### Grupos separados em: low, midlow, midhigh e high")

dfs <- extrair_diagonal_A(dados$A_list)
max_which_list <- nrow(dfs)
max_Bcr <- ncol(dfs)

# Data frame para armazenar todos os resultados
resultados_todos <- data.frame()

for (w in 1:max_which_list) {
  for (b in 1:max_Bcr) {
    
    # Rodar função uma vez, suprimindo mensagens e warnings
    res <- tryCatch({
      suppressWarnings(
        suppressMessages(
          km_quartis_exposure(
            E_list_medio   = E_list_medio,
            which_list     = w,
            Bcr            = b,
            dados_survival = dados_survival2
          )
        )
      )
    }, error = function(e) NULL)
    
    if (!is.null(res)) {
      
      p_formatado <- if (!is.null(res$p_value)) signif(res$p_value, 3) else NA
      
      # Armazena no data frame
      resultados_todos <- rbind(
        resultados_todos,
        data.frame(
          which_list = w,
          Bcr        = b,
          chisq      = if (!is.null(res$chisq)) res$chisq else NA,
          p_value    = if (!is.null(res$p_value)) res$p_value else NA
        )
      )
      
      cat("\n----------------------------------\n")
      cat("which_list =", w, "| Bcr =", b, "\n")
      cat("p-valor =", p_formatado, "\n")
     
      # Imprime gráfico com p-valor no título
      if (!is.null(res$plot)) {
        print(
          res$plot +
            ggplot2::ggtitle(
              paste0("Lista ", w,
                     " | Assinatura ", b,
                     " | p = ", p_formatado)
            )
        )
      }
    }
  }
}

# Ordena a tabela final por menor p
resultados_todos <- resultados_todos[
  order(resultados_todos$p_value),
]

print(resultados_todos)          
```

```r
# Filtra apenas os resultados significativos
resultados_significativos <- subset(resultados_todos, p_value < 0.05)

cat("\n\n#########################################\n")
cat("### Resultados significativos (p < 0.05) ###\n")
cat("#########################################\n\n")

print(resultados_significativos)
```

```r
km_low_high_exposure <- function(
  E_list_medio,
  which_list,
  Bcr,
  dados_survival
) {

  # ===== Exposição =====
  nomes_raw <- E_list_medio[[which_list]][Bcr, ]

  final_df <- data.frame(
    TCGA = sub(
      ".*(TCGA-[A-Za-z0-9]{2}-[A-Za-z0-9]{4}).*",
      "\\1",
      names(nomes_raw)
    ),
    exposure = as.numeric(nomes_raw),
    stringsAsFactors = FALSE
  )

  final_df <- final_df[!is.na(final_df$TCGA), ]

  # ===== Clínico =====
  dados <- dados_survival[
    !is.na(dados_survival$feature) &
    !is.na(dados_survival$time),
  ]

  extract_patient_barcode <- function(x) {
    pat <- "(TCGA-[A-Za-z0-9]{2}-[A-Za-z0-9]{4})"
    regmatches(x, regexpr(pat, x, perl = TRUE))
  }

  dados$TCGA <- extract_patient_barcode(dados$bcr_patient_barcode)

  # ===== Merge =====
  dados <- merge(dados, final_df, by = "TCGA", all = FALSE)

  # ===== Checagem =====
  if(length(unique(dados$exposure)) < 2){
    message("Sem variabilidade suficiente.")
    return(NULL)
  }

  # ===== LOW vs HIGH (mediana) =====
  mediana <- median(dados$exposure, na.rm = TRUE)

  dados$grupo <- ifelse(dados$exposure <= mediana, "low", "high")
  dados$grupo <- factor(dados$grupo, levels = c("low", "high"))

  # ===== Evento =====
  dados$event <- ifelse(
    tolower(dados$vital_status) == "dead",
    1, 0
  )

  # ===== KM =====
  surv_obj <- Surv(dados$time, dados$event)
  fit_km <- survfit(surv_obj ~ grupo, data = dados)

  plot(
    fit_km,
    col  = c("#2166ac", "#b2182b"),
    lwd  = 2,
    main = paste("KM LOW vs HIGH – Lista", which_list, "| Assinatura", Bcr)
  )

  legend(
    "bottomleft",
    legend = levels(dados$grupo),
    col    = c("#2166ac", "#b2182b"),
    lwd    = 2,
    bty    = "n"
  )

  # ===== Log-rank =====
  lr <- survdiff(surv_obj ~ grupo, data = dados)
  p_val <- 1 - pchisq(lr$chisq, 1)

  return(list(
    fit_km  = fit_km,
    chisq   = lr$chisq,
    p_value = p_val,
    dados   = dados
  ))
}
```

```r
res_km <- km_low_high_exposure(
  E_list_medio,
  which_list = 1,
  Bcr = 3,
  dados_survival = dados_survival2
)

if (!is.null(res_km)) {
  print(res_km$p_value)
}
```
